```python
orders_df = orders_df.dropDuplicates(["order_id"])
```

```python
orders_df = orders_df.na.drop(how="all")
```

```python
from pyspark.sql.functions import to_timestamp
orders_df = (
    orders_df
    .withColumn(
        "order_purchase_timestamp",
        to_timestamp("order_purchase_timestamp")
    )
    .withColumn(
        "order_approved_at",
        to_timestamp("order_approved_at")
    )
    .withColumn(
        "order_delivered_carrier_date",
        to_timestamp("order_delivered_carrier_date")
    )
    .withColumn(
        "order_delivered_customer_date",
        to_timestamp("order_delivered_customer_date")
    )
    .withColumn(
        "order_estimated_delivery_date",
        to_timestamp("order_estimated_delivery_date")
    )
)
```

```python
orders_df.select("order_status").distinct().show()
```

```python
from pyspark.sql.functions import col,isnan,when,count
orders_df.select(
    [
        count(
            when(col(c).isNull(), c)
        ).alias(c)
        for c in orders_df.columns
    ]
).show()
```

```python
from pyspark.sql.functions import *

customers_df = spark.read.table("BR_Customers")

sl_customers = (
    customers_df
    .dropDuplicates(["customer_id"])
    .na.drop(how="all")
    .withColumn("customer_city", initcap(trim(col("customer_city"))))
    .withColumn("customer_state", upper(trim(col("customer_state"))))
)

sl_customers.write.mode("overwrite").format("delta").saveAsTable("SL_Customers")
```

```python
from pyspark.sql.functions import *

items_df = spark.read.table("BR_OrderItems")

sl_items = (
    items_df
    .dropDuplicates()
    .na.drop(how="all")
    .withColumn("price", col("price").cast("double"))
    .withColumn("freight_value", col("freight_value").cast("double"))
    .filter(col("price") >= 0)
    .filter(col("freight_value") >= 0)
)

sl_items.write.mode("overwrite").format("delta").saveAsTable("SL_OrderItems")
```

```python
from pyspark.sql.functions import *

payments_df = spark.read.table("BR_Order_Payments")

sl_payments = (
    payments_df
    .dropDuplicates()
    .na.drop(how="all")
    .withColumn("payment_installments",
                col("payment_installments").cast("int"))
    .withColumn("payment_value",
                col("payment_value").cast("double"))
    .filter(col("payment_value") > 0)
)

sl_payments.write.mode("overwrite").format("delta").saveAsTable("SL_Order_Payments")
```

```python
from pyspark.sql.functions import *

reviews_df = spark.read.table("BR_Orders_Reviews")

sl_reviews = (
    reviews_df
    .dropDuplicates(["review_id"])
    .withColumn("review_score",
                col("review_score").cast("int"))
    .withColumn(
        "review_comment_title",
        trim(
            coalesce(col("review_comment_title"), lit("No Title"))
        )
    )
    .withColumn(
        "review_comment_message",
        trim(
            coalesce(col("review_comment_message"), lit("No Review"))
        )
    )
)

sl_reviews.write.mode("overwrite").format("delta").saveAsTable("SL_Order_Reviews")
```

```python
from pyspark.sql.functions import *

category_df = spark.read.table("BR_Product_Category")

sl_category = (
    category_df
    .dropDuplicates()
    .na.drop(how="all")
    .withColumn(
        "product_category_name",
        lower(trim(col("product_category_name")))
    )
    .withColumn(
        "product_category_name_english",
        initcap(trim(col("product_category_name_english")))
    )
)

sl_category.write.mode("overwrite").format("delta").saveAsTable("SL_Product_Category")
```
